```python
df.isnull().sum()
```

```python
df['age']=df['age'].fillna(df['age'].median())

```

```python
df['likes']=df['likes'].fillna(round(df['likes'].median()))
```

```python
df['comments']=df['comments'].fillna(df['comments'].median())
df['shares']=df['shares'].fillna(df['shares'].median())
```

```python
df['gender']=df['gender'].fillna(df['gender'].mode()[0])
```

```python
df['sentiment']=df['sentiment'].fillna(df['sentiment'].mode()[0])
```

```python
text_columns = [ 'gender','country','post_type','post_category', 'device_type','sentiment']
for col in text_columns:
    extra_spaces = df[col].dropna().astype(str).str.contains(r'^\s|\s$')
    print(col, ":", extra_spaces.sum())
```

```python
df[['likes', 'comments', 'shares']].describe()
```

```python
print("Negative likes:", (df['likes'] < 0).sum())
print("Negative comments:", (df['comments'] < 0).sum())
print("Negative shares:", (df['shares'] < 0).sum())
```

```python
df['hashtag_count'] = df['hashtags'].fillna('').apply(lambda x: sum(word.startswith('#') for word in str(x).split()))
```

# ❖ Task 5 — Statistical Analysis  

Compute descriptive stats for likes, comments, shares, watch_time, engagement_rate,

followers:

● Mean, median, mode


● Standard deviation, variance
● Percentiles

● (Optional) Skewness and kurtosis

```python
print("Mean:", df['engagement_score'].mean())
print("Median:", df['engagement_score'].median())
print("Mode:", df['engagement_score'].mode()[0])
```

```python
descriptive_stat_col=['likes','comments','shares','watch_time_sec','engagement_rate','follower_count']
statistical_summary = pd.DataFrame ({'Mean': df[descriptive_stat_col].mean(),
    'Median': df[descriptive_stat_col].median(),
    'Mode': df[descriptive_stat_col].mode().iloc[0],
    'Standard Deviation': df[descriptive_stat_col].std(),
    'Variance': df[descriptive_stat_col].var(),
    '25th Percentile': df[descriptive_stat_col].quantile(0.25),
    '50th Percentile': df[descriptive_stat_col].quantile(0.50),
    '75th Percentile': df[descriptive_stat_col].quantile(0.75),
    'Skewness': df[descriptive_stat_col].skew(),
    'Kurtosis': df[descriptive_stat_col].kurtosis()})

statistical_summary
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,6))
plt.scatter(df['impression_count'], df['likes'],alpha=0.5,color='purple')
plt.title('Likes vs Impressions',fontweight='bold')
plt.xlabel('Impressions')
plt.ylabel('Likes')
plt.show()
```

```python
d_engagement = df.groupby(df['posted_at'].dt.date)['engagement_score'].mean()
```

```python
plt.figure(figsize=(14,6))
plt.plot(d_engagement.index, d_engagement.values)
plt.title('Daily Engagement Trend', fontsize=14, fontweight='bold')
plt.xlabel('Date')
plt.ylabel('Average Engagement Score')
plt.xticks(rotation=45)
plt.tight_layout()
plt.grid(True,alpha=0.5)
plt.show()
```

```python
import seaborn as sns
```

```python
plt.figure(figsize=(8, 6))
sns.countplot(data=df, x='post_type',)
plt.title('Distribution of Post Types', fontweight='bold', fontsize=14)
plt.xlabel('Post Type')
plt.ylabel('Number of Posts')
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.barplot(data=df, x='post_category', y='likes', color='skyblue')
plt.title('Average Likes by Category', fontweight='bold', fontsize=14)
plt.xlabel('Post Category')
plt.ylabel('Average Likes')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(9, 5))
sns.violinplot(data=df, x='sentiment', y='follower_count', color='green')
plt.title('Distribution of Followers by Sentiment', fontweight='bold', fontsize=14)
plt.xlabel('Sentiment')
plt.ylabel('Follower Count')
plt.show()
```

```python
correlation = df.corr(numeric_only=True)
plt.figure(figsize=(12, 8))
sns.heatmap(correlation, annot=True, cmap='viridis', fmt='.2f')
plt.title('Correlation Heatmap', fontweight='bold', fontsize=14)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.swarmplot(data=df.sample(500), x='device_type', y='engagement_score', size=3)
plt.title('Engagement Score by Device Type', fontweight='bold', fontsize=14)
plt.xlabel('Device Type')
plt.ylabel('Engagement Score')
plt.grid(True,alpha=0.5)
plt.show()
```
